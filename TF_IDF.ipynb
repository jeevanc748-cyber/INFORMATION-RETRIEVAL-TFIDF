```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample documents
documents = [
    "information retrieval is the process of obtaining information",
    "tf idf is used in information retrieval systems",
    "cosine similarity measures similarity between vectors"
]

query = ["information retrieval using tf idf"]

# Step 1: Vectorize
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents + query)

# Step 2: Split document and query vectors
doc_vectors = tfidf_matrix[:-1]
query_vector = tfidf_matrix[-1]

# Step 3: Compute similarity
similarities = cosine_similarity(query_vector, doc_vectors)

# Step 4: Rank documents
for i, score in enumerate(similarities[0]):
    print(f"Document {i+1} similarity: {score:.4f}")
```
